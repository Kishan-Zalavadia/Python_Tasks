```python
evenNum = []
mulof_17 = []
userNumber = []

value = input("enter number to get thevalues : ")

# Simple logic 
for i in range(1000):
    if(i % 2 == 0 and i != 0):
        evenNum.append(i)
    if(i % 17 == 0):
        mulof_17.append(i)
    if(i > int(value)):
        userNumber.append(i)

        
print(evenNum)
print(mulof_17)
print(userNumber)
```
