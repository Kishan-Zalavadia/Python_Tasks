```python
print("write no of input you want")
val = input("Enter number")
print(val)
alluser = {}
print("enter the name:age in input field")
for i in range(int(val)):
    key_value = input("enter value " + str(i+1)+ " = ")
    results  = key_value.split(":")
    alluser[results[0]] = int(results[1])

Range_18to35 = []
Range_36to50 = []
Range_51to65 = []
Range_66to90 = []

for name,age in alluser.items():
    if 18 <= age <= 35:
        Range_18to35.append(name)
    elif 36 <= age <= 50:
        Range_36to50.append(name)
    elif 51 <= age <= 65:
        Range_51to65.append(name)
    elif 66 <= age <= 90:
        Range_66to90.append(name)
    
    
oldest_Person = max(alluser,key=alluser.get)
youngest_Person= min(alluser,key=alluser.get)

print(f"\nYoungest Person = {youngest_Person} with age {alluser[youngest_Person]}")
print(f"oldest Person = {oldest_Person} with age {alluser[oldest_Person]}\n")


print("person in range 18 to 35",Range_18to35)
print("person in range 36 to 50",Range_36to50)
print("person in range 51 to 65",Range_51to65)
print("person in range 66 to 90",Range_66to90)
```
