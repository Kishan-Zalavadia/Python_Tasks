```python
class book:
    def __init__(self,book_name,auth_name):
        self.name = book_name
        self.auth = auth_name
    def printbook(self):
        print("book name is" + self.name + " auther is "+self.auth)
```

```python
def create_student():
    name  = input("write name here ")
    luckNum = int(input("write luckey num here "))
    hobbieslist = input("write hobbies by comma sprated ")
    hobbies = hobbieslist.split(",")
    noofbooks = int(input("how many book you have been read "))
    books = []
    for i in range(noofbooks):
        bookname = input("enter book name"+ str(i+1) )
        authname = input("enter auther name"+ str(i+1) )
        b1  = book(bookname,authname)
        books.append(b1)
    s1 = student(name,luckNum,hobbies,books)
    return s1
```
